##  List

```python
# 1(a) Create a list of ages

age_list = [24, 25, 26, 27, 28]
print(age_list)
```

```python
# 1(b) Create a list of names

name_list = ["Anu", "Amal", "Diya", "Soniya", "Meera"]
print(name_list)
```

```python
# 2(a) Append the string "Yazhini" to name_list

name_list.append("Yazhini")
print(name_list)
```

```python
# 2(b) Insert 30 at index 2 in age_list

age_list.insert(2, 30)
print(age_list)
```

```python
# 2(c) Remove "Yazhini" from name_list

name_list.remove("Yazhini")
print(name_list)
```

```python
# 2(d) Pop the last element from age_list

age_list.pop()
print(age_list)
```

```python
# 2(e) Extend age_list with additional ages

age_list.extend([29, 30, 26])
print(age_list)
```

```python
# 2(f) Sort age_list in descending order

age_list.sort(reverse=True)
print(age_list)
```

```python
# 2(g) Find maximum age, minimum age and sum of ages

print(max(age_list))
print(min(age_list))
print(sum(age_list))
```

```python
# 3(a) Print the first element of name_list

print(name_list[0])
```

```python
# 3(b) Print the last element of name_list

print(name_list[-1])
```

```python
# 3(c) Print elements from index 2 to index 4

print(name_list[2:5])
```

```python
# 3(d) Print name_list in reverse order

print(name_list[::-1])
```

## Dictionary

```python
# (a) Create a dictionary of student marks

student_marks = {"Anu": 85,"Amal": 76,"Diya": 92,"Arun": 68,"Meera": 88}
print(student_marks)
```

```python
# (b) Access the mark of a student

print(student_marks["Anu"])
```

```python
# (c) Add Janani with a mark of 80

student_marks["Janani"] = 80
print(student_marks)
```

```python
# (d) Update a student's mark to 82

student_marks["Amal"] = 82
print(student_marks)
```

```python
# (e) Print keys, values and items

print(student_marks.keys())
print(student_marks.values())
print(student_marks.items())
```

```python
# 1(c) Create two sets

set1 = {1, 3, 5, 7, 9}
set2 = {2, 3, 5, 8, 10}
print(set1)
print(set2)
```

## Operators and Conditional Statements

```python
score = float(input("Enter your score (0 to 10): "))

if score < 0 or score > 10:
    print("Invalid score. Enter a score between 0 and 10.")

elif score > 7:
    print("Above Average: Excellent performance! Keep it up.")

elif score >= 4:
    print("Average: Good effort! Keep practicing, there's room for improvement.")

else:
    print("Below Average: Need to improve your performance. Consistent practice will lead to better results.")
```
