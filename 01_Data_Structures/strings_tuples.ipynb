## String Concatenation

```python
string1 = "Hello "
name = input("Enter your Name: ")
result = string1 + name
print(result)

string3 = ", welcome to Python programming"
result = result + string3
print(result)
```

## String Slicing and Indexing

```python
string = "Hello Zara, welcome to Python programming"

# a. First character
print(string[0])
```

```python
# b. Last character
print(string[-1])
```

```python
# c. First 5 characters
print(string[:5])
```

```python
# d. Last 11 characters
print(string[-11:])
```

```python
# e. Reverse the string
print(string[::-1])
```

```python
# f. Print the word Python
string[23:29]
```

## String Methods

```python
strM = "Python beginner tutorial"

# a. Convert to uppercase
print(strM.upper())
```

```python
# b. Convert to lowercase
print(strM.lower())
```

```python
# c. Capitalize
print(strM.capitalize())
```

```python
# d. Count occurrences of character 't'
print("Number of 't':", strM.count('t'))
```

```python
# e. Replace Python with Machine Learning
print(strM.replace("Python", "Machine Learning"))
```

## Tuples


```python
tuple1 = (10, 20, 30)
tuple2 = (40, 50, 60)

# a. Concatenate the two tuples
t_combine = tuple1 + tuple2
print(t_combine)
```

```python
# b. Repeat the elements 3 times
t_repeat = t_combine * 3
print(t_repeat)
```

```python
# c. Access the 3rd element
print(t_combine[2])
```

```python
# d. Access the first three elements
print(t_combine[:3])
```

```python
# e. Access the last three elements
print(t_combine[-3:])
```
